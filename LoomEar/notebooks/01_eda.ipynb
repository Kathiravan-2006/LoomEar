# LoomEar: 01. Exploratory Data Analysis
This notebook explores the acoustic signatures of industrial power looms under healthy baseline operations and physical fault modes.

```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure project root is in sys.path
sys.path.insert(0, str(Path.cwd().parent))

from src.loomear.data.loaders import load_dataset, load_metadata
from src.loomear.features.handcrafted import extract_features_for_sample
```

```python
metadata_path = Path.cwd().parent / 'data' / 'metadata.csv'
df = load_metadata(metadata_path)
df.head()
```

```python
samples = load_dataset(Path.cwd().parent / 'data' / 'raw' / 'synthetic', metadata_path)
print(f'Successfully loaded {len(samples)} audio samples.')
```

### Plot Feature Distributions Across Conditions

```python
feat_rows = []
for s in samples:
    f = extract_features_for_sample(s.audio, sample_rate=s.sample_rate)
    f['condition'] = s.metadata['condition']
    f['machine_id'] = s.metadata['machine_id']
    feat_rows.append(f)
feat_df = pd.DataFrame(feat_rows)

fig, ax = plt.subplots(figsize=(8, 4))
sns.boxplot(data=feat_df, x='condition', y='kurtosis', palette='Set2', ax=ax)
ax.set_title('Kurtosis by Machine Operational Condition')
plt.show()
```
